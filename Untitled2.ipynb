```python
from PIL import Image, ImageDraw

def overlay_solid_dashed_paths(image_path, expected, predicted, cell_size=80, save_path=None):
    """
    Overlay expected (green, solid) and predicted (red, dashed) paths on the same maze image.
    """
    move_map = {"U": (-1, 0), "D": (1, 0), "L": (0, -1), "R": (0, 1)}
    
    img = Image.open(image_path).convert("RGBA")
    overlay = Image.new("RGBA", img.size, (255,255,255,0))
    draw = ImageDraw.Draw(overlay)
    
    offset_amount = cell_size * 0.15  # perpendicular offset
    dash_length = 12  # length of dashes in pixels

    def draw_line_segments(x0, y0, x1, y1, color, dashed=False):
        if not dashed:
            draw.line([x0, y0, x1, y1], fill=color+(255,), width=6)
        else:
            # Draw dashed line as small segments
            from math import hypot
            total_len = hypot(x1-x0, y1-y0)
            dx = (x1 - x0) / total_len * dash_length
            dy = (y1 - y0) / total_len * dash_length
            num_segments = int(total_len // dash_length)
            for i in range(0, num_segments, 2):
                xs = x0 + dx * i
                ys = y0 + dy * i
                xe = x0 + dx * min(i+1, num_segments)
                ye = y0 + dy * min(i+1, num_segments)
                draw.line([xs, ys, xe, ye], fill=color+(255,), width=6)

    def draw_path(path, color, is_expected=True, dashed=False):
        rows = img.height // cell_size
        r, c = rows - 1, 0  # start bottom-left
        for move in path:
            dr, dc = move_map[move]
            r_next = r + dr
            c_next = c + dc

            # Perpendicular offset
            if move in ["L", "R"]:
                dy = -offset_amount if is_expected else offset_amount
                dx = 0
            else:
                dx = -offset_amount if is_expected else offset_amount
                dy = 0

            x0 = c * cell_size + cell_size/2 + dx
            y0 = r * cell_size + cell_size/2 + dy
            x1 = c_next * cell_size + cell_size/2 + dx
            y1 = r_next * cell_size + cell_size/2 + dy

            draw_line_segments(x0, y0, x1, y1, color, dashed=dashed)
            r, c = r_next, c_next

    # Expected path → solid green
    draw_path(expected, (0, 255, 0), is_expected=True, dashed=False)
    # Predicted path → dashed red
    draw_path(predicted, (255, 0, 0), is_expected=False, dashed=True)

    combined = Image.alpha_composite(img.convert("RGBA"), overlay)
    
    if save_path:
        combined.save(save_path)
    
    return combined
```
